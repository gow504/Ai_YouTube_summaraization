```python
from google.colab import userdata
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
```

Extract YouTube Transcript

```python
from youtube_transcript_api import YouTubeTranscriptApi

def get_transcript(video_id):
    transcript = YouTubeTranscriptApi().fetch(video_id)
    print(transcript)
    text = " ".join([item.text for item in transcript])
    return text
```

Gemini Setup


```python
from google import genai
client = genai.Client(api_key=GOOGLE_API_KEY)
```

Analyze / Summarize Function


```python
def analyze_video(transcript):
   prompt = f"""
   Analyze this YouTube video transcript and provide:

   1. Summary (short)
   2. Detailed explanation
   3. Key points (bullet list)
   4. Important insights
   5. 5 Questions & Answers

   Transcript:
   {transcript}
   """

   response = client.models.generate_content(
       model="gemini-2.5-flash-lite",
       contents=prompt
   )

   return response.text
```

Run

```python
# Fetching transcript specifying the language code available for this video
video_id = "MNv5R6Ss1YY"  # replace

try:
    # Since the default 'en' is not available, we specify languages=['hi', 'en']
    # to look for Hindi first, and fall back to English if needed.
    transcript = get_transcript(video_id)
except Exception as e:
    print(f"Retrying with explicit Hindi language list...")
    from youtube_transcript_api import YouTubeTranscriptApi
    raw_transcript = YouTubeTranscriptApi.get_transcript(video_id, languages=['hi', 'en'])
    transcript = " ".join([item['text'] for item in raw_transcript])

result = analyze_video(transcript)
print(result)
```

Chat Model

```python
def ask_question(transcript, question):
   prompt = f"""
   Answer the question based on this video transcript:

   Transcript:
   {transcript}

   Question:
   {question}
   """

   response = client.models.generate_content(
       model="gemini-2.5-flash-lite",
       contents=prompt
   )

   return response.text

# Example
print(ask_question(transcript, "What is javascript"))
```
